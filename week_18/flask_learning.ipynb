## 2. 第一個 Flask 應用程式
這是一個最基本的 Flask 應用程式，它會在網頁上顯示 "Hello, World!"。

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def hello_world():
    return 'Hello, World!'

if __name__ == '__main__':
    app.run(debug=True)
```

## 3. 執行應用程式
要執行這個應用程式，請將上面的程式碼儲存為一個 Python 檔案（例如 `app.py`），然後在終端機中執行 `python app.py`。接著在瀏覽器中開啟 `http://127.0.0.1:5000/`，你應該會看到 "Hello, World!"。

## 4. 路由 (Routing)
路由是用來將 URL 和函式綁定在一起的。例如，我們可以建立一個新的路由，讓它在使用者造訪 `/user/<username>` 時，顯示使用者的名稱。

```python
from flask import Flask

app = Flask(__name__)

@app.route('/')
def hello_world():
    return 'Hello, World!'

@app.route('/user/<username>')
def show_user_profile(username):
    return f'User {username}'

if __name__ == '__main__':
    app.run(debug=True)
```

## 5. 執行含有新路由的應用程式
同樣地，將上面的程式碼儲存並執行。現在，除了首頁之外，你還可以造訪 `http://127.0.0.1:5000/user/你的名字`，網頁上將會顯示 `User 你的名字`。
